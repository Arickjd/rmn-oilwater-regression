#### Limitação dos dados?

Um nível similar de erro foi encontrado em vários métodos testados, o que indica uma limitação dos próprios dados. No gráfico abaixo, plotamos com uma mesma cor todas as curvas (suavizadas) para um dado valor verdadeiro de sat_oil. Observamos que para valores próximos, há muita sobreposição entre os dados. Mesmo para valores distantes, na faixa de [0.0-0.2] os dados possuem alta superposição, dificultando seu discernimento -- como observado nos resultados finais. Entendemos que este fator indica que seriam necessários métodos consideravelmente mais aprofundados para avançar neste problema.
